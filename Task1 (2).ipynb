```python
import pandas as pd
```

```python
import numpy as np
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("student_habits_performance.csv")
```

```python
df.head()
```

```python
df.shape
```

```python
df.columns
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df = df.drop_duplicates()
```

```python
df.info()
```

```python
df["parental_education_level"].value_counts()
```

```python
df["parental_education_level"] = df["parental_education_level"].fillna(
    df["parental_education_level"].mode()[0]
)
```

```python
df.isnull().sum()
```

```python
df.describe()
```

```python
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
print("Missing values:", df.isnull().sum().sum())
print("Duplicate rows:", df.duplicated().sum())
```

```python
df["exam_score"].mean()
```

```python
print("Average Exam Score:", round(df["exam_score"].mean(), 2))
```

```python
print("Highest Score:", df["exam_score"].max())
print("Lowest Score:", df["exam_score"].min())
```

```python
df["gender"].value_counts()
```

```python
df["gender"].value_counts().plot(kind="bar")

plt.title("Number of Students by Gender")
plt.xlabel("Gender")
plt.ylabel("Number of Students")
plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.hist(df["exam_score"], bins=10)

plt.title("Distribution of Exam Scores")
plt.xlabel("Exam Score")
plt.ylabel("Number of Students")

plt.show()
```

```python
print(
    "Average Study Hours:",
    round(df["study_hours_per_day"].mean(), 2)
)
```

```python
plt.figure(figsize=(8,5))

plt.hist(df["study_hours_per_day"], bins=10)

plt.title("Study Hours per Day")
plt.xlabel("Study Hours")
plt.ylabel("Number of Students")

plt.show()
```

```python
print(
    "Average Attendance:",
    round(df["attendance_percentage"].mean(), 2)
)
```

```python
plt.figure(figsize=(8,5))

plt.hist(df["attendance_percentage"], bins=10)

plt.title("Student Attendance Distribution")
plt.xlabel("Attendance Percentage")
plt.ylabel("Number of Students")

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.scatter(
    df["attendance_percentage"],
    df["exam_score"]
)

plt.title("Attendance vs Exam Score")
plt.xlabel("Attendance Percentage")
plt.ylabel("Exam Score")

plt.show()
```

```python
correlation = df["attendance_percentage"].corr(
    df["exam_score"]
)

print("Correlation between Attendance and Exam Score:",
      round(correlation, 2))
```

```python
correlation_matrix = df.corr(numeric_only=True)

correlation_matrix["exam_score"].sort_values(
    ascending=False
)
```

```python
plt.figure(figsize=(10,6))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Matrix")
plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.scatter(
    df["study_hours_per_day"],
    df["exam_score"]
)

plt.title("Study Hours vs Exam Score")
plt.xlabel("Study Hours per Day")
plt.ylabel("Exam Score")

plt.show()
```

```python
study_correlation = df["study_hours_per_day"].corr(
    df["exam_score"]
)

print(
    "Correlation between Study Hours and Exam Score:",
    round(study_correlation, 2)
)
```

```python
print("===== FINAL STATISTICS =====")

print("Average Exam Score:",
      round(df["exam_score"].mean(), 2))

print("Highest Exam Score:",
      df["exam_score"].max())

print("Lowest Exam Score:",
      df["exam_score"].min())

print("Average Attendance:",
      round(df["attendance_percentage"].mean(), 2))

print("Average Study Hours:",
      round(df["study_hours_per_day"].mean(), 2))

print("Attendance-Score Correlation:",
      round(df["attendance_percentage"].corr(df["exam_score"]), 2))

print("Study Hours-Score Correlation:",
      round(df["study_hours_per_day"].corr(df["exam_score"]), 2))
```

```python
top_students = df.sort_values(
    by="exam_score",
    ascending=False
).head(10)

top_students
```
